```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
file = "/Users/fintangilman/CODE_PROJECTS/thesis_code_python/make_reward_plot/SAC_36.csv"
data = pd.read_csv(file)
data = data.drop(columns=['Wall time'], errors='ignore')
data
```

```python
# Make ascpect ratio 16:9
plt.figure(figsize=(12, 7))
plt.plot(data['Step']/100, data['Value'])
plt.xlabel('Episode')
plt.ylabel('Reward')
plt.title('Episode Reward Over Time')
plt.grid(True, alpha=0.3)
plt.savefig('reward_plot.png', dpi=300)
```
